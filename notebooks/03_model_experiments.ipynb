```python
import pandas as pd
```

```python
df_model = pd.read_csv("../data/raw/lebron_game_logs_model.csv")
```

```python
pd.to_datetime(df_model["GAME_DATE"])

# FEATURES
feature_cols = [
    "PTS_LAG_1", "PTS_AVG_5", "PTS_AVG_10",
    "REB_LAG_1", "REB_AVG_5", "REB_AVG_10",
    "AST_LAG_1", "AST_AVG_5", "AST_AVG_10",
    "MIN_LAG_1", "MIN_AVG_5", "MIN_AVG_10",
    "REST_DAYS", "HOME"
]

# TARGETS
target_cols = ["PTS", "REB", "AST", "MIN"]
```

```python
# CHRONOLOGICAL ORDER
df_model = df_model.sort_values("GAME_DATE").reset_index(drop=True)

# POINT OF SEPARATION
split_index = int(len(df_model) * 0.8)

# TRAIN DATA (FIRST 80%)
X_train = df_model[feature_cols].iloc[:split_index]
y_train = df_model[target_cols].iloc[:split_index]

# TEST DATA (LAST 20%)
X_test = df_model[feature_cols].iloc[split_index:]
y_test = df_model[target_cols].iloc[split_index:]


print(f"Train: {X_train.shape}, {y_train.shape}")
print(f"Test: {X_test.shape}, {y_test.shape}")

print(f"\nPeriod of train:")
print(f"{df_model["GAME_DATE"].iloc[0]} until {df_model["GAME_DATE"].iloc[split_index - 1]}")

print(f"\nPeriod of test:")
print(f"{df_model["GAME_DATE"].iloc[split_index]} until {df_model["GAME_DATE"].iloc[-1]}")
```

```python
# VERIFICATIONS
assert X_train.index.max() < X_test.index.min()
assert not X_train.isna().any().any()
assert not X_test.isna().any().any()

print("Timing separation successfully validated")
```
