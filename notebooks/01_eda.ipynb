```python
import pandas as pd
```

```python
df = pd.read_csv("../data/raw/lebron_game_logs.csv")
```

```python
print(f"Shape: {df.shape}")
print(f"\nHead: \n{df.head()}")
print(f"\nDuplicate values: {df.duplicated()}")
print(f"\nData types: \n{df.dtypes}")
print(f"\nMissing values: \n{df.isna().sum()}")
print(f"\nSeasons: \n{df["SEASON_ID"].value_counts().sort_index()}")
```

```python
# TRANSFORMING GAME DATE TO DATETIME (CSV's GLITCH)
print(pd.to_datetime(df["GAME_DATE"]))
print(df["GAME_DATE"].dtype)
```

```python
# AVERAGES ANDS PCTs

df_avg = pd.DataFrame({
    "PPG": [df["PTS"].mean().round(1)],
    "APG": [df["AST"].mean().round(1)],
    "RPG": [df["REB"].mean().round(1)],
    "SPG": [df["STL"].mean().round(1)],
    "BPG": [df["BLK"].mean().round(1)],
    "FG3M_PG": [df["FG3M"].mean().round(1)],
    "FTA_PG": [df["FTA"].mean().round(1)],
    "FTM_PG": [df["FTM"].mean().round(1)],
    "FG_PCT": [df["FG_PCT"].mean().round(2)],
    "FG3_PCT": [df["FG3_PCT"].mean().round(2)],
    "FT_PCT": [df["FT_PCT"].mean().round(2)]
})

df_avg
```
